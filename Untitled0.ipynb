# Hotel Booking Demand - Data Pre-processing & Visualisation Pipeline
This notebook implements an end-to-end data cleaning, engineering, and visual intelligence pipeline for the hotel bookings dataset.

## Section 1: Data Ingestion
I load the raw CSV data into a Pandas DataFrame and check its presence in the environment.

```python
try:
    if not os.path.exists(DATASET_PATH):
        raise FileNotFoundError(f"Target dataset file '{DATASET_PATH}' not found in the current directory.")

    print("[INFO] Initiating dataset ingestion...")
    df_raw = pd.read_csv(DATASET_PATH)
    print(f"[INFO] Ingested raw dataset successfully. Initial shape: {df_raw.shape}")

    # Create a deep copy for processing to maintain data isolation
    df = df_raw.copy()
except Exception as e:
    print(f"[ERROR] Failed to load data: {str(e)}")
```

## Section 2: Data Cleaning & Quality Remediation
In this section, I programmatically address four data quality issues.

```python
try:
    # 2.1: Exact Duplicate Removal
    duplicate_count = df.duplicated().sum()
    if duplicate_count > 0:
        df.drop_duplicates(inplace=True)
        print(f"[CLEAN] Removed {duplicate_count} exact duplicate rows.")

    # 2.2: Missing Value Imputation
    print("[CLEAN] Addressing missing values across columns...")
    # 'children': Mode imputation (0)
    df['children'] = df['children'].fillna(0).astype(int)
    # 'country': Flag missing entries as 'Unknown'
    df['country'] = df['country'].fillna('Unknown')
    # 'agent' & 'company': 0 implies a direct booking without booking managers or companies
    df['agent'] = df['agent'].fillna(0).astype(int)
    df['company'] = df['company'].fillna(0).astype(int)

    # 2.3: Operational Constraints (0 total guests)
    zero_guest_mask = (df['adults'] == 0) & (df['children'] == 0) & (df['babies'] == 0)
    invalid_guest_count = zero_guest_mask.sum()
    if invalid_guest_count > 0:
        df = df[~zero_guest_mask]
        print(f"[CLEAN] Filtered out {invalid_guest_count} structurally invalid bookings (0 total guests).")

    # 2.4: Outlier Treatment (Erroneous ADR values)
    adr_outlier_mask = (df['adr'] < 0) | (df['adr'] > 1000)
    outlier_adr_count = adr_outlier_mask.sum()
    if outlier_adr_count > 0:
        df = df[~adr_outlier_mask]
        print(f"[CLEAN] Purged {outlier_adr_count} severe ADR outliers (negative values or values > $1000).")

    print(f"[INFO] Cleaning complete. Intermediate shape: {df.shape}")
except Exception as e:
    print(f"[ERROR] Failed during data cleaning phase: {str(e)}")
```

## Section 3: Feature Engineering & Formatting
I standardize categorical strings (`meal`), fix data types (`reservation_status_date`), and calculate metrics (`total_stay_nights`, `lead_time_segment`) to enhance our analysis.

```python
try:
    print("[FEATURE] Formatting features and parsing datetimes...")

    # Convert dates to datetime objects
    df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'], errors='coerce')

    # Standardize 'Undefined' meal entries to 'SC' (Self-Catering)
    df['meal'] = df['meal'].replace('Undefined', 'SC')

    # Calculate Total Stay Duration
    df['total_stay_nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']

    # Segment continuous Lead Times into operational planning blocks
    df['lead_time_segment'] = pd.cut(
        df['lead_time'],
        bins=[-1, 30, 90, 180, 1000],
        labels=['0-30 Days', '31-90 Days', '91-180 Days', '181+ Days']
    )

    # Save processed copy to disk
    df.to_csv(OUTPUT_PATH, index=False)
    print(f"[SUCCESS] Processed dataset saved locally as '{OUTPUT_PATH}'. Shape: {df.shape}")
except Exception as e:
    print(f"[ERROR] Failed during formatting and feature engineering: {str(e)}")
```

## Section 4: Data Visualisation - Simple Business Questions
I construct three visual metrics to answer macro-level operational questions.

Question 1: What is the
proportional distribution of bookings across hotel types? (Pie Chart)

```python
try:
    # --- Question 1: Booking Distribution by Hotel Type ---
    fig1, ax1 = plt.subplots(figsize=(6, 6))
    hotel_counts = df['hotel'].value_counts()
    ax1.pie(
        hotel_counts, labels=hotel_counts.index, autopct='%1.1f%%',
        startangle=90, colors=['#4c72b0', '#dd8452'], explode=(0, 0.05),
        textprops={'fontsize': 11}
    )
    ax1.set_title('Distribution of Bookings by Hotel Type', fontsize=14, fontweight='bold', pad=15)
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"[ERROR] Failed to render simple visualizations: {str(e)}")
```

Question 2: Which meal plans are most heavily preferred by guests? (Sorted Bar Chart)

```python
try:
  # --- Question 2: Preferred Meal Types ---
    fig2, ax2 = plt.subplots(figsize=(8, 5))
    meal_counts = df['meal'].value_counts().sort_values(ascending=False)
    sns.barplot(x=meal_counts.index, y=meal_counts.values, ax=ax2, palette='viridis', hue=meal_counts.index, dodge=False)
    ax2.set_title('Preferred Meal Types by Guests (Sorted)', fontsize=14, fontweight='bold', pad=12)
    ax2.set_xlabel('Meal Categorization Code')
    ax2.set_ylabel('Number of Bookings')
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"[ERROR] Failed to render simple visualizations: {str(e)}")
```

Question 3: What is the volume contribution of individual market segments? (Sorted Bar Chart)

```python
try:
   # --- Question 3: Bookings by Market Segment ---
    fig3, ax3 = plt.subplots(figsize=(10, 5))
    market_counts = df['market_segment'].value_counts().sort_values(ascending=False)
    sns.barplot(x=market_counts.index, y=market_counts.values, ax=ax3, palette='magma', hue=market_counts.index, dodge=False)
    ax3.set_title('Bookings volume by Market Segment (Sorted)', fontsize=14, fontweight='bold', pad=12)
    ax3.set_xlabel('Distribution / Market Segment Channel')
    ax3.set_ylabel('Number of Bookings')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"[ERROR] Failed to render simple visualizations: {str(e)}")
```

## Section 5: Data Visualisation - Complex Business Questions
I deploy multi-dimensional, advanced charts to unpack complex behavioral dynamics.

Question 4: How do continuous variables linearly interact with cancellation behavior? (Heatmap Matrix)

```python
try:
    # --- Question 4: Correlation Matrix Heatmap ---
    fig4, ax4 = plt.subplots(figsize=(10, 8))
    numeric_cols = [
        'is_canceled', 'lead_time', 'adr', 'total_stay_nights',
        'booking_changes', 'total_of_special_requests',
        'required_car_parking_spaces', 'adults', 'children'
    ]
    corr_matrix = df[numeric_cols].corr()
    sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', ax=ax4, square=True, linewidths=.5, cbar_kws={'shrink': 0.8})
    ax4.set_title('Correlation Heatmap of Numerical Features & Cancellation Status', fontsize=14, fontweight='bold', pad=15)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"[ERROR] Failed to render complex visualizations: {str(e)}")
```

Question 5: What do seasonal monthly ADR fluctuations look like across the two hotel types? (Line Trend Plot)

```python
try:
   # --- Question 5: Monthly ADR Fluctuations ---
    month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
    df['arrival_date_month'] = pd.Categorical(df['arrival_date_month'], categories=month_order, ordered=True)
    monthly_adr = df.groupby(['arrival_date_month', 'hotel'], observed=False)['adr'].mean().reset_index()

    fig5, ax5 = plt.subplots(figsize=(12, 6))
    sns.lineplot(data=monthly_adr, x='arrival_date_month', y='adr', hue='hotel', marker='o', ax=ax5, linewidth=2.5, palette=['#4c72b0', '#dd8452'])
    ax5.set_title('Average Daily Rate (ADR) Fluctuation Across Months', fontsize=14, fontweight='bold', pad=12)
    ax5.set_xlabel('Month of Arrival')
    ax5.set_ylabel('Mean Average Daily Rate ($)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"[ERROR] Failed to render complex visualizations: {str(e)}")
```

Question 6: How do lead times and customer profiles interact to dictate cancellation rates? (Grouped Segment Bar Chart)

```python
try:
   # --- Question 6: Cancellation Rate by Customer Type and Lead Time Segment ---
    cancel_rates = df.groupby(['lead_time_segment', 'customer_type'], observed=False)['is_canceled'].mean().reset_index()

    fig6, ax6 = plt.subplots(figsize=(12, 6))
    sns.barplot(data=cancel_rates, x='customer_type', y='is_canceled', hue='lead_time_segment', ax=ax6, palette='Blues_d')
    ax6.set_title('Cancellation Rate by Customer Type and Lead Time Segment', fontsize=14, fontweight='bold', pad=12)
    ax6.set_xlabel('Customer Profile Categorization')
    ax6.set_ylabel('Cancellation Rate (Proportional Metric)')
    ax6.set_ylim(0, 1.0)
    plt.tight_layout()
    plt.show()

    print("[SUCCESS] Full visual intelligence pipeline executed successfully.")

except Exception as e:
    print(f"[ERROR] Failed to render complex visualizations: {str(e)}")
```
